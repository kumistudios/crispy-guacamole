```python
# Generate synthetic datasets (Ghana & India), run preprocessing, clustering (GMM, Spectral, DBSCAN/HDBSCAN fallback),
# UMAP/TSNE for visualization, construct similarity graph and spectral embeddings as proxy for GraphSAGE,
# and produce Figures 1-5 and Tables 1-4.
import numpy as np, pandas as pd, matplotlib.pyplot as plt, os, textwrap
from sklearn.mixture import GaussianMixture
from sklearn.cluster import SpectralClustering, DBSCAN, AgglomerativeClustering
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.manifold import TSNE
from sklearn.decomposition import TruncatedSVD
from sklearn.neighbors import kneighbors_graph
from sklearn.manifold import SpectralEmbedding
import networkx as nx
import warnings
warnings.filterwarnings("ignore")

np.random.seed(42)
```

```python
# 1) Create synthetic datasets
def make_synthetic_country(name, n=2000):
    ages = np.random.choice([18,22,27,35,45,55,65], size=n, p=[0.15,0.2,0.2,0.2,0.12,0.08,0.05])
    genders = np.random.choice(['Male','Female'], size=n, p=[0.52,0.48])
    employment = np.random.choice(['Employed','Unemployed','Student','Retired'], size=n, p=[0.6,0.15,0.15,0.1])
    data_used = np.random.gamma(2.0, 2.0, size=n) * (2.0 if name=='India' else 1.2)
    call_minutes = np.random.gamma(2.5, 5.0, size=n) * (1.1 if name=='Ghana' else 1.0)
    monthly_charges = 5 + 0.5*data_used + 0.02*call_minutes + np.random.randn(n)*2 + (3.0 if name=='India' else 1.0)
    churn = (np.random.rand(n) < (0.12 if name=='India' else 0.18)).astype(int)
    telecom_partner = np.random.choice(['TelcoA','TelcoB','TelcoC'], size=n, p=[0.5,0.3,0.2])
    return pd.DataFrame({
        'country': name,
        'age': ages,
        'gender': genders,
        'employment': employment,
        'data_used': data_used,
        'call_minutes': call_minutes,
        'monthly_charges': monthly_charges,
        'churn': churn,
        'TelecomCompany': telecom_partner
    })
ghana = make_synthetic_country('Ghana', n=1500)
india = make_synthetic_country('India', n=2500)
df = pd.concat([ghana, india], ignore_index=True)
print("Datasets created:", df.shape)
```

```python
# 2) Preprocessing per methodology
df['AgeGroup'] = pd.cut(df['age'], bins=[17,24,34,44,54,64,100], labels=['18-24','25-34','35-44','45-54','55-64','65+'])
# Feature engineering
df['avg_data_per_min'] = df['data_used'] / (df['call_minutes'] + 1e-6)
df['behavior_score'] = (1 - df['churn']) * np.log1p(df['data_used']) + 0.01*df['call_minutes']
# Encode categorical
le_gender = LabelEncoder().fit(df['gender'])
df['gender_enc'] = le_gender.transform(df['gender'])
le_emp = LabelEncoder().fit(df['employment'])
df['emp_enc'] = le_emp.transform(df['employment'])
le_tel = LabelEncoder().fit(df['TelecomCompany'])
df['tel_enc'] = le_tel.transform(df['TelecomCompany'])
features = ['data_used','call_minutes','monthly_charges','avg_data_per_min','behavior_score','gender_enc','emp_enc','tel_enc']
X = df[features].fillna(0).values
scaler = StandardScaler().fit(X)
Xs = scaler.transform(X)
```

```python
# 3) Clustering: GMM, Spectral, DBSCAN/HDBSCAN fallback
# GMM
gmm = GaussianMixture(n_components=5, random_state=42).fit(Xs)
gmm_labels = gmm.predict(Xs)
# Spectral
try:
    spectral = SpectralClustering(n_clusters=5, affinity='nearest_neighbors', random_state=42).fit(Xs)
    spectral_labels = spectral.labels_
except Exception as e:
    spectral_labels = AgglomerativeClustering(n_clusters=5).fit_predict(Xs)
# DBSCAN (density)
db = DBSCAN(eps=1.2, min_samples=15).fit(Xs)
db_labels = db.labels_

df['gmm_cluster'] = gmm_labels
df['spectral_cluster'] = spectral_labels
df['dbscan_cluster'] = db_labels
```

```python
# 4) Dimensionality reduction for visualization: UMAP if available else TSNE
try:
    import umap
    reducer = umap.UMAP(n_components=2, random_state=42)
    X2 = reducer.fit_transform(Xs)
    method_used = 'UMAP'
except Exception as e:
    tsne = TSNE(n_components=2, random_state=42, init='pca', learning_rate='auto', perplexity=30)
    X2 = tsne.fit_transform(Xs)
    method_used = 't-SNE'

df['x2'] = X2[:,0]
df['y2'] = X2[:,1]
```

```python
# 5) Cluster evaluation metrics for GMM and Spectral (excluding noise labels)
sil_gmm = silhouette_score(Xs, gmm_labels)
sil_spectral = silhouette_score(Xs, spectral_labels)
# For DBSCAN, if one cluster only, handle
try:
    if len(set(db_labels))>1:
        sil_db = silhouette_score(Xs, db_labels)
    else:
        sil_db = np.nan
except Exception:
    sil_db = np.nan
```

```python
# ARI between methods
ari_gmm_spectral = adjusted_rand_score(gmm_labels, spectral_labels)

# Table 1: Summary statistics by country
def summary_stats(dfc):
    n = len(dfc)
    return pd.Series({
        'n': n,
        'mean_age': dfc['age'].mean(),
        'prop_female': (dfc['gender']=='Female').mean(),
        'mean_data_used': dfc['data_used'].mean(),
        'mean_call_minutes': dfc['call_minutes'].mean(),
        'mean_monthly_charges': dfc['monthly_charges'].mean(),
        'churn_rate': dfc['churn'].mean()
    })

table1 = pd.concat([summary_stats(df[df['country']=='Ghana']), summary_stats(df[df['country']=='India'])], axis=1).T.reset_index().rename(columns={'index':'country'})
```

```python
# Table 2: Cluster evaluation results
table2 = pd.DataFrame({
    'method': ['GMM','Spectral','DBSCAN'],
    'silhouette': [sil_gmm, sil_spectral, sil_db],
    'notes': ['probabilistic','graph-based spectral','density-based (noise allowed)']
})
```

```python
# 6) Graph construction: kNN graph based on features + cluster edges
A = kneighbors_graph(Xs, n_neighbors=10, mode='connectivity', include_self=False)
G = nx.from_scipy_sparse_array(A)
# Add cluster membership as node attribute
for i,node in enumerate(G.nodes()):
    G.nodes[node]['country'] = df.iloc[node]['country']
    G.nodes[node]['gmm'] = int(df.iloc[node]['gmm_cluster'])
# Spectral embedding on graph as proxy for GraphSAGE embeddings
spec_emb = SpectralEmbedding(n_components=16, affinity='precomputed')
A_dense = A.toarray()
# For spectral embedding, need symmetric adjacency; use A_dense @ A_dense.T to ensure connectivity
aff = (A_dense + A_dense.T)/2
emb = spec_emb.fit_transform(aff + np.eye(aff.shape[0])*1e-6)
# Use embedding to build simple recommender: nearest neighbor similarity in embedding space
from sklearn.neighbors import NearestNeighbors
nn = NearestNeighbors(n_neighbors=6).fit(emb)
distances, indices = nn.kneighbors(emb)
# For evaluation, we'll create a synthetic ground truth: treat users with same TelecomCompany and same country as positives
# Evaluate Precision@5 averaged
def precision_at_k(indices, df, k=5):
    precisions = []
    for i,row in enumerate(df.itertuples()):
        neigh = indices[i,1:k+1]  # exclude self at position 0
        true_vals = ((df.iloc[neigh]['TelecomCompany']==row.TelecomCompany) & (df.iloc[neigh]['country']==row.country)).astype(int)
        precisions.append(true_vals.sum()/k)
    return np.mean(precisions)

prec5 = precision_at_k(indices, df, k=5)
```

```python
# Table 3: Model performance metrics
table3 = pd.DataFrame({
    'model': ['GMM+SpectralEmb Recomm (proxy)'],
    'precision_at_5': [prec5],
    'notes': ['spectral-embedding proxy for GraphSAGE; similarity-based recommender']
})
```
